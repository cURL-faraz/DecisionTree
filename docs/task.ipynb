# Task 3

Overfitting: The model learns both patterns and noise in the training data, performing well on training but poorly on unseen data.

Underfitting: The model is too simple to capture the patterns, performing poorly on both training and test data.

When the model has learned the noise, it means the decision boundaries are fitted to random fluctuations or irrelevant patterns in the training data, reducing generalization.

## Decision Tree Hyperparameters and Their Effect:
	•	max_depth:
	•	Too large → overfitting (deep, complex tree).
	•	Too small → underfitting (shallow tree).
	•	min_samples_split:
	•	Too small → overfitting (splits on tiny, noisy subsets).
	•	Too large → underfitting (few splits, tree too simple).
	•	min_samples_leaf:
	•	Too small → overfitting (leaves with very few samples).
	•	Too large → underfitting (loss of detail).
	•	max_features:
	•	Too high → overfitting (too much info per split).
	•	Too low → underfitting (insufficient info per split).

## A greedy decision tree (splitting only for immediate best gain) might not be globally optimal. To get an optimal decision tree:
	•	Use pruning (post-pruning or pre-pruning) to remove unnecessary branches.
	•	Apply exhaustive search or dynamic programming for small datasets (computationally expensive).
	•	Use meta-heuristic algorithms (e.g., genetic algorithms, simulated annealing) to search for better trees.
	•	Ensemble methods like Random Forest or Gradient Boosted Trees can approximate a better structure.
